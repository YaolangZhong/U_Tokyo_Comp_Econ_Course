## 2. Predict the answer

We convert degrees Celsius, $C$, to degrees Fahrenheit, $F$, using

$$F = C \times \frac{9}{5} + 32.$$

For $C=20$, predict the result before running the code. The input cell below stores the temperature under the name `celsius`. It does not print anything by itself.


```python
celsius = 20.0
```

## 3. Run the calculation

This cell uses the current value of `celsius`, calculates `fahrenheit`, and prints both values to one decimal place. The baseline result is **20.0 C = 68.0 F**.


```python
fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius:.1f} C = {fahrenheit:.1f} F")
```

## 5. Check two familiar temperatures

The following cell checks the same conversion rule at 0 and 100 degrees Celsius. An `assert` checks whether a statement is true; if it is false, Python reports an error.


```python
assert 0 * 9 / 5 + 32 == 32
assert 100 * 9 / 5 + 32 == 212
print("Both checks passed: 0 C = 32 F and 100 C = 212 F.")
```
